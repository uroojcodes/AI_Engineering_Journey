```python
import requests

new_appointment = {
    "title": "Appointment with Urooj",
    "date": "2026-10-10",
    "time": "14:00"
}

response = requests.post("https://jsonplaceholder.typicode.com/posts", json=new_appointment)

print(response.status_code)
print(response.json())
```
